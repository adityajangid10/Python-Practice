```python
# Subplot()

# The subplot function takes 3 arguments that describe the layout of the figure.

# The layout is organized in rows and columns, which are represented by the first and second arguments.

# The third argument represents the index of the current plot.

# Example:
# plt.subplot(1,2,1)

# Here the figure has 1 row, 2 columns, and this plot is the first plot, as the last argument is 1.

# plt.subplot(1,2,2)

# Here the figure has 1 row, 2 columns, and this plot is the second plot, as the last argument is 2.
```

```python
import matplotlib.pyplot as plt
```

```python
# plot -1
student = ['A','B','C','D','E','f']
height = [163,145,167,180,178,177]

plt.plot(student,height)

plt.show()
```

```python
# plot -2
student = ['A','B','C','D','E','f']
weight = [173,135,157,190,158,177]

plt.plot(student,weight)

plt.show()
```

```python
# Data
students = [1, 2, 3, 4, 5]

height = [150,170 , 165,160 , 155]
weight = [163,145,167,180,178]
marks = [60, 70, 80, 75, 90]
age = [18, 15, 20, 18, 18]
plt.suptitle("Student statistics") # super title --> suptitle
# 1st Graph - Height
plt.subplot(2, 2, 1)
plt.plot(students, height)
plt.title("Student Height")
plt.xlabel("Student")
plt.ylabel("Height")

# 2nd Graph - Weight
plt.subplot(2, 2, 2)
plt.plot(students, weight)
plt.title("Student Weight")
plt.xlabel("Student")
plt.ylabel("Weight")

# 3rd Graph - Marks
plt.subplot(2, 2, 3)
plt.plot(students, marks)
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")

# 4th Graph - Age
plt.subplot(2, 2, 4)
plt.plot(students, age)
plt.title("Student Age")
plt.xlabel("Student")
plt.ylabel("Age")

plt.tight_layout()
plt.show()
```

```python
students = [1, 2, 3, 4, 5]

height = [150, 155, 160, 165, 170]
weight = [45, 50, 55, 60, 65]
marks = [60, 70, 80, 75, 90]
age = [18, 19, 20, 21, 22]
attendance = [80, 85, 75, 90, 95]
study_hours = [2, 3, 4, 5, 6]


# Graph 1
plt.subplot(3, 2, 1)
plt.plot(students, height)
plt.title("Student Height")

# Graph 2
plt.subplot(3, 2, 2)
plt.plot(students, weight)
plt.title("Student Weight")

# Graph 3
plt.subplot(3, 2, 3)
plt.plot(students, marks)
plt.title("Student Marks")

# Graph 4
plt.subplot(3, 2, 4)
plt.plot(students, age)
plt.title("Student Age")

# Graph 5
plt.subplot(3, 2, 5)
plt.plot(students, attendance)
plt.title("Attendance")

# Graph 6
plt.subplot(3, 2, 6)
plt.plot(students, study_hours)
plt.title("Study Hours")

plt.tight_layout()
plt.show()
```
